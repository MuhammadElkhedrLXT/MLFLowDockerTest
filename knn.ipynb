```python
import mlflow

# The set_experiment API creates a new experiment if it doesn't exist.
mlflow.set_tracking_uri("http://localhost:5000")
mlflow.set_experiment("KNN Tuning")
```

```python
from sklearn.model_selection import train_test_split
import sklearn.datasets 

X, y = sklearn.datasets.load_iris(return_X_y=True)
X_train, X_val, y_train, y_val = train_test_split(X, y, random_state=42, test_size= 0.2)

print(X_train)
print(y_train)
```

```python
from sklearn.model_selection import ParameterGrid, cross_val_score, GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

parameter_grid = {'n_neighbors': [i for i in range(1 , 16)], "weights": ["uniform" , "distance"]}
mlflow.sklearn.autolog(max_tuning_runs=None)
with mlflow.start_run(run_name="GridSearchCV tuning"):
    knn = KNeighborsClassifier()
    grid_search = GridSearchCV(knn, param_grid=parameter_grid, cv = 5, scoring = "accuracy", n_jobs = -1)
    grid_search.fit(X_train, y_train)
    
    best_score = grid_search.score(X_val, y_val)
    
    mlflow.log_param("best_params", grid_search.best_params_)
    mlflow.log_metric("best_test_score", best_score)
```
